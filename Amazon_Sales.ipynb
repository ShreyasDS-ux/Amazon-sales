```python
# Part 12: Visualizations 
# Bar Charts 
# 1.	Revenue by Product Category. 
Revenue = df.groupby("product_category")["total_revenue"].sum().sort_values(ascending=False)
Revenue.plot(kind='bar')
plt.title("Revenue by Product Category")
plt.ylabel("Revenue")
plt.xlabel("Product Category")
plt.show()
```

```python
# 2.	Revenue by Region.
revenue = df.groupby("customer_region")["total_revenue"].sum().sort_values(ascending=False).head()
revenue.plot(kind='bar')
plt.title("Revenue by Region")
plt.ylabel("Revenue")
plt.xlabel("Region")
plt.show()
```

```python
# 3.	Revenue by Payment Method.  
Payment = df.groupby("payment_method")["total_revenue"].sum().sort_values(ascending=False).head()
Payment.plot(kind='bar')
plt.title("Revenue by Payment Method")
plt.ylabel("Revenue")
plt.xlabel("Payment Method")
plt.show()
```

```python
# Horizontal Bar Charts 
# 1.	Top 10 Products by Revenue.  
product = df.groupby("product_id")["total_revenue"].sum().sort_values().head(10)
product.plot(kind='barh')
plt.title("Top 10 Products by Revenue")
plt.ylabel("Revenue")
plt.xlabel("Product_id")
plt.show()
```

```python
# Pie Charts 
# 1.	Revenue Contribution by Category.  
cont=df["product_category"].value_counts()
m=cont.index
n=cont.values
plt.title("Revenue on catogery")
a=["maroon", "purple", "pink", "red","yellow","Green"]
plt.pie(n,labels=m,colors=a)
plt.legend(n)
plt.show()
```

```python
# 2.	Orders by Payment Method.  
Payment = df["payment_method"].value_counts()
Z = Payment.index
X = Payment.values
plt.title("Orders by Payment Method")
a=["Green", "yellow", "pink", "red","orange","blue"]
plt.pie(X,labels=Z,colors=a)
plt.legend(X)
plt.show()
```

```python
# Line Charts 
# 1.	Monthly Revenue Trend.
Month = df.loc[1:200, "month_name"]
Revenue = df.loc[1:200,"total_revenue"]
plt.figure(figsize=(20,10),dpi=100)
plt.plot(Month,Revenue,linestyle='--',color='b',marker= 'D')
plt.title("Monthly Revenue Trend")
plt.ylabel("Revenue")
plt.xlabel("Month",color='b', size=8, weight='bold',style='italic',family= 'monospace')
plt.show()
```

```python
# 2.	Monthly Sales Trend.
Month = df.loc[:200,"month_name"]
Sales = df.loc[:200,"quantity_sold"]
plt.figure(figsize=(20,10),dpi=100)
plt.plot(Month,Sales,linestyle='--',color='b',marker= 'D')
plt.xlabel("Months")
plt.ylabel("Quantity")
plt.show()
```

```python
# 3.	Daily Revenue Trend 
Day =df.loc[:200, "day_name"]
Revenue = df.loc[:200,"total_revenue"]
plt.figure(figsize=(20,10),dpi=100)
plt.plot(Day,Revenue,linestyle='--',color ='b',marker='D')
plt.title("Daily Revenue Trend")
plt.ylabel("Revenue")
plt.xlabel("Day")
plt.show()
```

```python
# 4.	Daily Sales Trend 
Day = df.loc[:100,"day_name"]
Sales = df.loc[:100,"quantity_sold"]
plt.figure(figsize=(20,10),dpi=100)
plt.plot(Day,Sales,linestyle='--',marker='D',color='b')
plt.title("Daily Qty sold")
plt.ylabel("Qty sales")
plt.xlabel("Day")
plt.show()
```
